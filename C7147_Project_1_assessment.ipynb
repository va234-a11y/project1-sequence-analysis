## 🧬 Load reference and reads data into variables
Below, we provide some helper code to load the reference from `reference.txt` and the reads from `reads.csv`.

❗ **Do not change the function’s behavior** (you may read it, of course!).

The code creates two variables from the loaded files:
- **reference**: a string containing the 200-nt reference sequence
- **reads**: a list of dictionaries containing the 6,000 reads

````python
##### Constants

REF_PATH = "reference.txt"
READS_PATH = "reads.csv"

##### Functions

def load_reference(path: str) -> str:
    """Load reference nucleotide sequence from file at path to reference string."""
    with open(path) as f:
        lines = [ln.strip() for ln in f if ln.strip()]
    if lines[0].startswith('>'):
        lines = lines[1:]
    ref = "".join(lines).upper()
    if len(ref) != 200:
        raise ValueError(f"Reference must be 200 nt, got {len(ref)}")
    return ref


def load_reads(path: str) -> list:
    """Load sample NGS reads from file at path to reads list."""
    with open(path, newline="") as f:
        reader = csv.DictReader(f)
        reads = list(reader)
    return reads

##### Main code

# Load reference and reads
reference = load_reference(REF_PATH)
reads = load_reads(READS_PATH)

# Report data structures
print("Reference length:", len(reference))
print("Total no. of reads:", len(reads))
print("Keys in each read dictionary:", reads[0].keys())
print("Samples loaded:", sorted({read["sample"] for read in reads}))

# Print sample sequences
print("\nSample read data:")
for row in reads[:5]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
print("...")
for row in reads[-5:]:
    print(row["read_id"], row["sample"], row['group'], row["sequence"])
````

## Step 1 — Validate each read and record any errors
## Function: `validate_read(sequence, reference)`

### 👍 You can assume:
- `reference` is a non-empty string containing only uppercase A/C/G/T.
- `sequence` is a string, but nothing about its length or contents.

### 🎯 Aims:
- Check that an individual read is the same length as the reference and contains only valid characters (A/C/G/T)
- "Repair" any reads that contain valid lower-case characters -> upper-case
- Return a dictionary with the following keys:

    - `sequence`: the upper-case version of the sequence input, including when invalid.
    - `valid`: `True` only when the length matches the reference and every character is A/C/G/T.
    - `errors`: a list of strings. Include `"wrong_length"` first if lengths differ, then `"invalid_characters"` if any character is outside A/C/G/T. Include both when both apply. A valid read has `[]`.

### 🔎 Expected outputs

Examples for reference `"ATGC"`:
| Input (read) | Input type | Output |
|---|---|:---|
| `"ATGC"` | Correct | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `"atgc"` | Lowercase | `{"sequence": "ATGC", "valid": True, "errors": []}` |
| `""` | Empty read | `{"sequence": "", "valid": False, "errors": ["wrong_length"]}` |
| `"ATNC"` | Invalid character | `{"sequence": "ATNC", "valid": False, "errors": ["invalid_characters"]}` |
| `"ATN"` | Multiple errors | `{"sequence": "ATN", "valid": False, "errors": ["wrong_length", "invalid_characters"]}` |


Invalid reads should not crash your program.

### 💻 Enter your code here

````python
#def validate_read(sequence, reference):
    #"""Replace this with a docstring describing the function."""

 # Write your code here.

def validate_read(sequence, reference):
    """Checks if sequence is same length as reference and only contains A,T,G,C."""
    upper_sequence= sequence.upper() #changing any lowercase characters to uppercase
    errors= []

    if len(upper_sequence) != len(reference): #checking the length of sequence
      errors.append("wrong_length")

    if not all(base in "ATGC" for base in upper_sequence): #ensuring there are only correct characters
      errors.append("invalid_characters")

    return {
        "sequence":upper_sequence,
        "valid": len(errors)== 0,
        "errors": errors}
````

## Step 2 — Find every mismatch in a valid read
## Function: `find_mutations(sequence, reference)`

### 👍 You can assume:
- Inputs `sequence` and `reference` are valid (i.e. both are strings of the same length and contain only upper-case bases, i.e. A/C/G/T)

### 🎯 Aims:
- Return a list containing one dictionary per mismatch, in ascending position order.
- Each dictionary has exactly these keys:
    - `position`: position of the mutation (integer), numbered from 1.
    - `reference_base`: one-character string from reference at that position.
    - `read_base`: one-character string from the read at that position.

- Return `[]` if the read matches the reference.

### 🔎 Expected outputs

Example: `find_mutations("GTGA", "ATGC")` returns:
```python
[{"position": 1, "reference_base": "A", "read_base": "G"},
 {"position": 4, "reference_base": "C", "read_base": "A"}]
```


````python
def find_mutations(sequence, reference):
    """Finds position where sequence is different from reference."""
    mutation= []
    for i, base in enumerate(sequence): #checks index
      if base != reference[i]:   #checks for any differences in bases
        mutation.append({ #adds dictionary with details of mutation
            "position": i + 1,
            "reference_base": reference[i],
            "read_base": base
            })

    return mutation
````

## Step 3 — Analyse all reads
## Function: `analyse_reads(reads, reference)`

### 👍 You can assume:

- `reference` is a non-empty string containing only A/C/G/T.

### 🎯 Aims:
- Use `validate_read` and `find_mutations` to return a **new list of dictionaries containing information about all reads** (even invalid ones, one dictionary per read), preserving the input read order.
- Each read dictionary should contain the following keys (values below are examples):
```python
{
    "read_id": "read_0001",
    "sample": "sample1",
    "group": "treatment",
    "sequence": "GTGA",             # normalised upper-case sequence
    "status": "valid",              # "valid" or "invalid"
    "errors": [],                   # validation error codes
    "mutations": [
        {"position": 1, "reference_base": "A", "read_base": "G"},
        {"position": 4, "reference_base": "C", "read_base": "A"}
    ]
}
```
- Do not change the original read dictionaries.

### 🔎 Expected outputs
- Valid reads have `status="valid"` and `mutations` listing all mutations in the dictionary format shown above.
- Invalid reads have `status="invalid"`, the relevant error codes, and `mutations=[]`.
- Empty input returns `[]`.

### 💻 Enter your code here

````python
def analyse_reads(reads, reference):
    """Analyses all reads and identifies mutations of valid reads."""
    results = []
    for r in reads:
        check = validate_read(r["sequence"], reference)

        if check["valid"] == True:
            status = "valid"
            mutations = find_mutations(check["sequence"], reference)
        else:
            status = "invalid"
            mutations = []

        read_result = {
            "read_id": r["read_id"],
            "sample": r["sample"],
            "group": r["group"],
            "sequence": check["sequence"],
            "status": status,
            "errors": check["errors"],
            "mutations": mutations
        }
        results.append(read_result)
    return results
````

````python
#creating a new dictionary for each sample to store its information


def new_sample_record(group):
  """Returns an empty summary dictionary for sample"""
  return {
          "group": group,
          "valid_reads": 0,
          "invalid_reads": 0,
          "mismatch_reads": 0,
          "mismatch_percentage": None,
          "mismatch_events":0,
          "substitution_counts": {},
          "position_counts": {},
          "top_positions": []}




# counts how often each substitution or position occurs
def add_one(counts, key):
  "Add 1 to counts[key] and starts with 1 when there is new key"
  counts[key]= counts.get(key, 0) + 1


#calculate the percentage
def calculate_percentage(count, total):
  """Gives the count as a percentage of a total."""
  if total == 0:
    return None
  return count/ total * 100

#identifying the three positions where the most amount fo mismatches occur
def find_top_positions(position_counts, n_top= 3):
    """Returns the specified number of positions with the most mismatches."""
    ranked_positions = sorted(
        position_counts.items(),
        key=lambda item: (-item[1], item[0]))
    return [position for position, count in ranked_positions][:n_top]


#only the valid reads are checked for mismatches and count each affected read once
def count_read(record, read):
  """Updates the sample record from the read and adds a count."""
  if read["status"] != "valid":
        record["invalid_reads"] +=1
        return

  record["valid_reads"] += 1
  mutations = read["mutations"]

  if mutations:
    record["mismatch_reads"] += 1

  for mutation in mutations:
    record["mismatch_events"] += 1
    substitution = mutation["reference_base"] + ">" + mutation["read_base"]
    add_one(record["substitution_counts"], substitution)
    add_one(record["position_counts"], mutation["position"])

def final_record(record):
  """ Calculates percentages and identifies top positions."""
  record["mismatch_percentage"] = calculate_percentage(record["mismatch_reads"], record["valid_reads"])

#sorting the dictionaries by keys
  record["substitution_counts"] = dict(sorted(record["substitution_counts"].items()))
  record["position_counts"] = dict(sorted(record["position_counts"].items()))
  record["top_positions"] = find_top_positions(record["position_counts"])



#organising the sample names by the experimental group and prints the results separately
def group_samples(sample_summary):
  """Returns the group with the sample names sorted."""
  groups={}
  for sample in sorted(sample_summary):
    groups.setdefault(sample_summary[sample]["group"],[]).append(sample)
  return groups





# shows the overview of read quality and mismatch information for each sample
def print_results_table(results, samples):
    """Print one row of read and mismatch results per sample."""
    print("\nSummary of Reads and Mismatches")
    header = (
        f"{'Sample':<15}"
        f"{'Valid reads':>12}"
        f"{'Invalid reads':>14}"
        f"{'Mismatch reads (%)':>22}"
        f"{'Mismatch events':>18}"
        f"{'Top 3 positions':>20}")
    print(header)
    print(" -" * len(header))

    for sample in samples:
        record = results[sample]

        if record["mismatch_percentage"] is None:
            percentage = "N/A"
        else:
            percentage = f"{record['mismatch_percentage']:.1f}%"


#calculating top positon
        top_pos = ", ".join(str(p) for p in record["top_positions"]) or "none"

        print(
            f"{sample:<15}"
            f"{record['valid_reads']:>12}"
            f"{record['invalid_reads']:>14}"
            f"{percentage:>22}"
            f"{record['mismatch_events']:>18}"
            f"{top_pos:>20}")



def print_substitution_table(results, samples):
  """Print subsitutional counts and percentage for every 100 valid reads."""
  print("\nSubstitutional Mutations")
  print("Counts and percentage per 100 valid reads")

  substitutions = sorted({substitution
        for sample in samples
        for substitution in results[sample]["substitution_counts"]})

  header = f"{'Substitution':<18}" + "".join(f"{sample:>22}" for sample in samples)
  print( header)
  print("-" * len(header))

  if not substitutions:
        print("No substitutions observed.")
        return

  for substitution in substitutions:
        row = f"{substitution:<18}"
        for sample in samples:
            record = results [sample]
            count = record["substitution_counts"].get(substitution, 0)
            percentage = calculate_percentage(count, record["valid_reads"])
            if percentage is None:
                value = f"{count} (N/A)"
            else:
                value = f"{count} ({percentage:.1f}%)"
            row += f"{value:>22}"
        print(row)


def print_group_tables(results):
    """Print a results table and a substitution table for each group."""
    if not results:
        print("No reads to summarise.")
        return

    for group, samples in group_samples(results).items():
        print(f"\n{'=' * 100}")
        print(f"GROUP: {group}")
        print(f"{'=' * 100}")
        print_results_table(results, samples)
        print_substitution_table(results, samples)


def summarise_samples(analysed_reads):
  """Summarises read validity and mismatch patterns for each sample.
  Prints summary table and substitution table for each group and pre-sample results."""

  results={}

  for read in analysed_reads:
        sample = read["sample"]
        if sample not in results:
            results[sample] = new_sample_record(read["group"])
        count_read(results[sample], read)

  for record in results.values():
        final_record(record)

  print_group_tables(results)
  return results
````

### 📊 Statistical comparison

For each position, pass the sample percentages to the supplied function as a list (one per group) and store the result, e.g.:

```python
statistical_result = calculate_statistics(control_percentages, treated_percentages)
```

Each list must contain **one percentage per biological sample**.

The supplied function returns:

```python
{"statistic": float, "p_value": float}
```

It returns `None` values when the comparison cannot be calculated.

````python
#@title 📊 calculate_statistics function

from scipy import stats
import math
import warnings


def calculate_statistics(control_percentages, treated_percentages):
    """Compare independent biological samples using a two-sided Welch's t-test."""
    unavailable = {"statistic": None, "p_value": None}

    # Require at least two samples per group and finite numerical values.
    if len(control_percentages) < 2 or len(treated_percentages) < 2:
        return unavailable

    values = list(control_percentages) + list(treated_percentages)
    if any(value is None or not math.isfinite(value) for value in values):
        return unavailable

    # No within-group variation: the t-test cannot estimate uncertainty.
    if (
        len(set(control_percentages)) == 1
        and len(set(treated_percentages)) == 1
    ):
        return unavailable

    with warnings.catch_warnings():
        warnings.simplefilter("ignore", RuntimeWarning)
        result = stats.ttest_ind(
            treated_percentages,
            control_percentages,
            equal_var=False,
            alternative="two-sided"
        )

    if not (
        math.isfinite(result.statistic)
        and math.isfinite(result.pvalue)
    ):
        return unavailable

    return {
        "statistic": float(result.statistic),
        "p_value": float(result.pvalue)
    }
````
